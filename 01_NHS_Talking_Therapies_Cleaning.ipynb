## Data cleaning

* Raw data was imported from six separate CSV files (2019/20–2024/25).

* Original dataset: [NHS Talking Therapies, for anxiety and depression, Annual reports](https://digital.nhs.uk/data-and-information/publications/statistical/nhs-talking-therapies-for-anxiety-and-depression-annual-reports)

* Due to inconsistent column naming across years, each year's data was standardised using DuckDB SQL queries before concatenating into a single dataset.

* Suppressed values (marked as * in the original files) were replaced with NaN, and numeric columns were converted to the appropriate data types.

* The cleaned dataset was saved as a single CSV for subsequent analysis.

```python
import duckdb
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')

path = "/content/drive/MyDrive/NHS_data_analysis_project/"

file_names = [
    "psych-ther-ann-rep-csv-2019-20-main-csv.csv",
    "psych-ther-ann-rep-csv-2020-21-main-csv-v2.csv",
    "psych-ther-ann-rep-csv-2021-22-main-csv.csv",
    "psych-ther-ann-rep-csv-2022-23-main-v2.csv",
    "psych-ther-ann-rep-csv-2023-24-main.csv",
    "psych-ther-ann-rep-csv-2024-25-main2.csv"
]

dfs = []
for name in file_names:
    full_path = path + name
    print(f"importing: {name}")
    dfs.append(pd.read_csv(full_path))

df1, df2, df3, df4, df5, df6 = dfs[0], dfs[1], dfs[2], dfs[3], dfs[4], dfs[5]
print(df1.info())
print(df2.info())
print(df3.info())
print(df4.info())
print(df5.info())
print(df6.info())
```

```python
sql_query = """
SELECT
'2019/20' AS Year,
OrgType, OrgCode, OrgName,
VariableType, VariableA, VariableB,
CountReferralsReceived AS Count_ReferralsReceived,
CountFirstTreatment AS Count_AccessingServices,
CountFinishedCourseTreatment AS Count_FinishedCourseTreatment,
PercentageFinishedCourseTreatment AS Percentage_FinishedCourseTreatment,
CountEndedReferrals AS Count_EndedReferrals,
CountEndedBeforeTreatment AS Count_EndedBeforeTreatment,
MeanWaitEnteredTreatment AS Mean_WaitAccessingServices,
MeanWaitFinishedCourseTreatment AS Mean_WaitFinishedCourseTreatment,
PercentageImprovement AS Percentage_Improvement,
PercentageNoReliableChange AS Percentage_NoReliableChange,
PercentageDeterioration AS Percentage_Deterioration,
PercentageRecovery AS Percentage_Recovery,
PercentageReliableRecovery AS Percentage_ReliableRecovery,
MeanTreatmentAppointmentsRecovery AS Mean_TreatmentAppointmentsRecovery,
MeanTreatmentAppointmentsImprovement AS Mean_TreatmentAppointmentsImprovement,
MeanTreatmentAppointmentsNoChange AS Mean_TreatmentAppointmentsNoChange,
MeanTreatmentAppointmentsDeterioration AS Mean_TreatmentAppointmentsDeterioration,
MeanTreatmentAppointmentsReliableRecovery AS Mean_TreatmentAppointmentsReliableRecovery
FROM df1
"""
df1 = duckdb.query(sql_query).to_df()
df1.shape
```

```python
sql_query = """
SELECT
'2020/21' AS Year,
OrgType, OrgCode, OrgName,
VariableType, VariableA, VariableB,
Count_ReferralsReceived,
Count_FirstTreatment AS Count_AccessingServices,
Count_FinishedCourseTreatment,
Percentage_FinishedCourseTreatment,
Count_EndedReferrals,
Count_EndedBeforeTreatment,
Mean_WaitEnteredTreatment AS Mean_WaitAccessingServices,
Mean_WaitFinishedCourseTreatment,
Percentage_Improvement,
Percentage_NoReliableChange,
Percentage_Deterioration,
Percentage_Recovery,
Percentage_ReliableRecovery,
Mean_TreatmentAppointmentsRecovery,
Mean_TreatmentAppointmentsImprovement,
Mean_TreatmentAppointmentsNoChange,
Mean_TreatmentAppointmentsDeterioration,
Mean_TreatmentAppointmentsReliableRecovery
FROM df2
"""
df2 = duckdb.query(sql_query).to_df()
df2.shape
```

```python
import numpy as np

df3 = df3.replace('*',np.nan)

sql_query = """
SELECT
'2021/22' AS Year,
OrgType, OrgCode, OrgName,
VariableType, VariableA, VariableB,
Count_ReferralsReceived,
Count_FirstTreatment AS Count_AccessingServices,
Count_FinishedCourseTreatment,
Percentage_FinishedCourseTreatment,
Count_EndedReferrals,
Count_EndedBeforeTreatment,
Mean_WaitEnteredTreatment AS Mean_WaitAccessingServices,
Mean_WaitFinishedCourseTreatment,
Percentage_Improvement,
Percentage_NoReliableChange,
Percentage_Deterioration,
Percentage_Recovery,
Percentage_ReliableRecovery,
Mean_TreatmentAppointmentsRecovery,
Mean_TreatmentAppointmentsImprovement,
Mean_TreatmentAppointmentsNoChange,
Mean_TreatmentAppointmentsDeterioration,
Mean_TreatmentAppointmentsReliableRecovery
FROM df3
"""
df3 = duckdb.query(sql_query).to_df()
df3.shape
```

```python
df4 = df4.replace('*',np.nan)

sql_query = """
SELECT
'2022/23' AS Year,
OrgType, OrgCode, OrgName,
VariableType, VariableA, VariableB,
Count_ReferralsReceived,
Count_AccessingServices AS Count_AccessingServices,
Count_FinishedCourseTreatment,
Percentage_FinishedCourseTreatment,
Count_EndedReferrals,
Count_EndedBeforeTreatment,
Mean_WaitAccessingServices AS Mean_WaitAccessingServices,
Mean_WaitFinishedCourseTreatment,
Percentage_Improvement,
Percentage_NoReliableChange,
Percentage_Deterioration,
Percentage_Recovery,
Percentage_ReliableRecovery,
Mean_TreatmentAppointmentsRecovery,
Mean_TreatmentAppointmentsImprovement,
Mean_TreatmentAppointmentsNoChange,
Mean_TreatmentAppointmentsDeterioration,
Mean_TreatmentAppointmentsReliableRecovery
FROM df4
"""
df4 = duckdb.query(sql_query).to_df()
df4.shape
```

```python
df5 = df5.replace('*',np.nan)

sql_query = """
SELECT
'2023/24' AS Year,
OrgType, OrgCode, OrgName,
VariableType, VariableA, VariableB,
Count_ReferralsReceived,
Count_AccessingServices AS Count_AccessingServices,
Count_FinishedCourseTreatment,
Percentage_FinishedCourseTreatment,
Count_EndedReferrals,
Count_EndedBeforeTreatment,
Mean_WaitEnteredTreatment AS Mean_WaitAccessingServices,
Mean_WaitFinishedCourseTreatment,
Percentage_Improvement,
Percentage_NoReliableChange,
Percentage_Deterioration,
Percentage_Recovery,
Percentage_ReliableRecovery,
Mean_TreatmentAppointmentsRecovery,
Mean_TreatmentAppointmentsImprovement,
Mean_TreatmentAppointmentsNoChange,
Mean_TreatmentAppointmentsDeterioration,
Mean_TreatmentAppointmentsReliableRecovery
FROM df5
"""
df5 = duckdb.query(sql_query).to_df()
df5.shape
```

```python
df6 = df6.replace('*',np.nan)

sql_query = """
SELECT
'2024/25' AS Year,
OrgType, OrgCode, OrgName,
VariableType, VariableA, VariableB,
Count_ReferralsReceived,
Count_AccessingServices,
Count_FinishedCourseTreatment,
Percentage_Variable_withFCT AS Percentage_FinishedCourseTreatment,
Count_EndedReferrals,
Count_EndedBeforeTreatment,
Mean_WaitAccessingServices,
Mean_WaitFinishedCourseTreatment,
Percentage_Improvement,
Percentage_NoReliableChange,
Percentage_Deterioration,
Percentage_Recovery,
Percentage_ReliableRecovery,
Mean_TreatmentAppointmentsRecovery,
Mean_TreatmentAppointmentsImprovement,
Mean_TreatmentAppointmentsNoChange,
Mean_TreatmentAppointmentsDeterioration,
Mean_TreatmentAppointmentsReliableRecovery
FROM df6
"""
df6 = duckdb.query(sql_query).to_df()
df6.shape
```

```python
df_all = pd.concat([df1, df2, df3, df4, df5, df6], ignore_index=True)
df_all.info()
```

```python
int_columns = [
    'Count_ReferralsReceived', 'Count_AccessingServices',
    'Count_FinishedCourseTreatment', 'Count_EndedReferrals',
    'Count_EndedBeforeTreatment'
]

float_columns = [
    'Percentage_FinishedCourseTreatment',
    'Mean_WaitAccessingServices', 'Mean_WaitFinishedCourseTreatment',
    'Percentage_Improvement', 'Percentage_NoReliableChange',
    'Percentage_Deterioration', 'Percentage_Recovery', 'Percentage_ReliableRecovery',
    'Mean_TreatmentAppointmentsRecovery', 'Mean_TreatmentAppointmentsImprovement',
    'Mean_TreatmentAppointmentsNoChange', 'Mean_TreatmentAppointmentsDeterioration',
    'Mean_TreatmentAppointmentsReliableRecovery'
]


for col in int_columns:
    if col in df_all.columns:
        df_all[col] = pd.to_numeric(df_all[col], errors='coerce').astype('Int64')

for col in float_columns:
    if col in df_all.columns:
        df_all[col] = pd.to_numeric(df_all[col], errors='coerce').astype('float64')

df_all.info()
```

```python
save_path = path + "NHS_Talking_Therapies_Cleaned_All_Years.csv"
df_all.to_csv(save_path, index=False)
print(f"Saved to: {save_path}")
```
